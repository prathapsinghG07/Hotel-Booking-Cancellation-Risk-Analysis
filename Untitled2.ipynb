# Hotel Booking Cancellation Risk Analysis

### 1. Project Objective

Problem statement:
Build a machine learning classification system that predicts whether a hotel booking will be cancelled, allowing hotels to identify high-risk bookings and take preventive actions.

Target: is_canceled

Business value:

Identify high-risk reservations.
Reduce revenue loss from cancellations.
Improve room inventory planning.
Support targeted customer retention strategies.

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings

warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)
```

```python
# load dataset
df = pd.read_csv("/content/hotel_booking.csv")

print("Shape:", df.shape)
display(df.head())
```

image.png

```python
# Basic info
print(df.info())

print("\nMissing values:")
print(df.isnull().sum().sort_values(ascending=False).head(15))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nTarget distribution:")
print(df["is_canceled"].value_counts())
```

```python
drop_cols = [
    "reservation_status",
    "reservation_status_date",
    "name",
    "email",
    "phone-number",
    "credit_card"
]

df = df.drop(columns=drop_cols)

print("New shape:", df.shape)
```

### Feature engineering

```python
df["total_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

df["total_guests"] = (
    df["adults"] +
    df["children"].fillna(0) +
    df["babies"]
)

df["total_previous_bookings"] = (
    df["previous_cancellations"] +
    df["previous_bookings_not_canceled"]
)

df["is_family"] = (
    (df["children"].fillna(0) + df["babies"]) > 0
).astype(int)

print(df.shape)
```

```python
# Data cleaning
df = df[df["adults"] > 0]

print("Shape after cleaning:", df.shape)
```

### EDA

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="is_canceled")
plt.title("Booking Cancellation Distribution")
plt.show()

plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="is_canceled", y="lead_time")
plt.title("Lead Time vs Cancellation")
plt.show()

plt.figure(figsize=(8,5))
sns.barplot(
    data=df,
    x="market_segment",
    y="is_canceled"
)
plt.xticks(rotation=45)
plt.title("Cancellation Rate by Market Segment")
plt.show()
```

The exploratory analysis reveals meaningful variation in hotel booking cancellation behavior. The target distribution shows that cancellations account for approximately 37% of reservations, making this a relevant binary classification problem. Lead time and market segment demonstrate useful relationships with cancellation behavior, while the missing-value analysis highlights company and agent as important preprocessing considerations.

Overall, the EDA provides evidence that booking timing, booking channel, customer information, reservation characteristics, and historical booking behavior can be useful predictors of cancellation risk. These findings provide the foundation for the subsequent feature engineering and machine learning stages.

###Train test spliting

```python
X = df.drop("is_canceled", axis=1)
y = df["is_canceled"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing :", X_test.shape)
```

```python
numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical:", len(numeric_features))
print("Categorical:", len(categorical_features))
```

### preprocessing pipeline

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=True
        ))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

### Evaluation function

```python
def evaluate_model(model, X_test, y_test, model_name):

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    print("\n" + "="*50)
    print(model_name)
    print("="*50)

    print("Accuracy :", round(accuracy_score(y_test, y_pred), 4))
    print("Precision:", round(precision_score(y_test, y_pred), 4))
    print("Recall   :", round(recall_score(y_test, y_pred), 4))
    print("F1 Score :", round(f1_score(y_test, y_pred), 4))
    print("ROC-AUC  :", round(roc_auc_score(y_test, y_prob), 4))

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        y_pred
    )

    plt.title(model_name)
    plt.show()

    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    }
```

```python
#Logistic Regression
baseline_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            class_weight="balanced"
        ))
    ]
)

baseline_model.fit(X_train, y_train)

baseline_result = evaluate_model(
    baseline_model,
    X_test,
    y_test,
    "Logistic Regression - Baseline"
)
```

```python
#Random Forest
rf_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(
            n_estimators=300,
            max_depth=15,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ))
    ]
)

rf_model.fit(X_train, y_train)

rf_result = evaluate_model(
    rf_model,
    X_test,
    y_test,
    "Random Forest"
)
```

###XGBoost model

```python

from xgboost import XGBClassifier

xgb_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", XGBClassifier(
            n_estimators=300,
            max_depth=5,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            min_child_weight=3,
            reg_alpha=0.1,
            reg_lambda=1,
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42,
            n_jobs=-1
        ))
    ]
)

xgb_model.fit(X_train, y_train)

xgb_result = evaluate_model(
    xgb_model,
    X_test,
    y_test,
    "XGBoost"
)
```

```python
# XGBoost hyperparameter tuning
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    "classifier__n_estimators": [200, 300, 500, 700],

    "classifier__max_depth": [3, 4, 5, 6],

    "classifier__learning_rate": [
        0.01, 0.03, 0.05, 0.08, 0.1
    ],

    "classifier__subsample": [
        0.7, 0.8, 0.9, 1.0
    ],

    "classifier__colsample_bytree": [
        0.7, 0.8, 0.9, 1.0
    ],

    "classifier__min_child_weight": [
        1, 3, 5, 7
    ],

    "classifier__gamma": [
        0, 0.1, 0.3, 0.5
    ],

    "classifier__reg_alpha": [
        0, 0.01, 0.1, 0.5
    ],

    "classifier__reg_lambda": [
        1, 1.5, 2, 3
    ]
}
```

```python
#Randomized Search
xgb_tuning = RandomizedSearchCV(
    estimator=xgb_model,
    param_distributions=param_grid,
    n_iter=30,
    scoring="roc_auc",
    cv=3,
    verbose=2,
    random_state=42,
    n_jobs=-1
)

xgb_tuning.fit(X_train, y_train)
```

```python
#Best Parameters
print("Best Parameters:")
print(xgb_tuning.best_params_)

print("\nBest Cross-Validation ROC-AUC:")
print(round(xgb_tuning.best_score_, 4))
```

```python
#Evaluate Tuned XGBoost
tuned_xgb = xgb_tuning.best_estimator_

tuned_xgb_result = evaluate_model(
    tuned_xgb,
    X_test,
    y_test,
    "Tuned XGBoost"
)
```

### Compare All Models

```python
results = pd.DataFrame([
    baseline_result,
    rf_result,
    xgb_result,
    tuned_xgb_result
])

results = results.sort_values(
    by="ROC-AUC",
    ascending=False
).reset_index(drop=True)

display(results)
```

## Select Final Model

```python
best_model = tuned_xgb

print("Final model: Tuned XGBoost")
```

### Feature Importance

```python
preprocessor_fitted = (
    tuned_xgb.named_steps["preprocessor"]
)

classifier = (
    tuned_xgb.named_steps["classifier"]
)

feature_names = (
    preprocessor_fitted
    .get_feature_names_out()
)

importance = classifier.feature_importances_

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

display(
    feature_importance.head(15)
)
```

```python
#Plot top features
plt.figure(figsize=(8, 6))

sns.barplot(
    data=feature_importance.head(15),
    x="Importance",
    y="Feature"
)

plt.title("Top 15 Important Features - Tuned XGBoost")
plt.tight_layout()
plt.show()
```

### Final Confusion Matrix

```python
y_pred = best_model.predict(X_test)

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Final Model")

plt.show()
```

### Final Classification Report

```python
print(
    classification_report(
        y_test,
        y_pred
    )
)
```
